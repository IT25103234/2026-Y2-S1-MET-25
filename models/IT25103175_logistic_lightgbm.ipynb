```python
# 1. Import libraries
import pandas as pd
import numpy as np

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)
```

```python
from google.colab import files

uploaded = files.upload()
```

```python
# ============================================================
# 2. Load preprocessed train and test data
# ============================================================

train_path = "/content/final_train_processed.csv"
test_path = "/content/final_test_processed.csv"

train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

print("Training data shape:", train_df.shape)
print("Testing data shape:", test_df.shape)
```

```python
# ============================================================
# 3. Separate features (X) and target (y)
# ============================================================

X_train = train_df.drop("Exited", axis=1)
y_train = train_df["Exited"]

X_test = test_df.drop("Exited", axis=1)
y_test = test_df["Exited"]

print("\nX_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
```

```python
# ============================================================
# 4. Create Logistic Regression model
# ============================================================

logistic_model = LogisticRegression(
    max_iter=1000,
    class_weight="balanced",
    random_state=42
)
```

```python
# ============================================================
# 5. Train the model
# ============================================================

logistic_model.fit(X_train, y_train)

print("\nLogistic Regression model trained successfully.")
```

```python
# ============================================================
# 6. Make predictions
# ============================================================

y_pred = logistic_model.predict(X_test)

# Probability of churn
y_prob = logistic_model.predict_proba(X_test)[:, 1]
```

```python
# ============================================================
# 7. Calculate evaluation metrics
# ============================================================

accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)
roc_auc = roc_auc_score(y_test, y_prob)

print("\n========== LOGISTIC REGRESSION RESULTS ==========")

print(f"Accuracy  : {accuracy:.4f}")
print(f"Precision : {precision:.4f}")
print(f"Recall    : {recall:.4f}")
print(f"F1 Score  : {f1:.4f}")
print(f"ROC-AUC   : {roc_auc:.4f}")
```

```python
# ============================================================
# 8. Confusion Matrix
# ============================================================

cm = confusion_matrix(y_test, y_pred)

print("\nConfusion Matrix:")
print(cm)
```

```python
# ============================================================
# 9. Classification Report
# ============================================================

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred,
    target_names=["Retained", "Churned"]
))
```

```python
# 1. Import libraries
import pandas as pd
import numpy as np

import lightgbm as lgb

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)
```

```python

# ============================================================
# 2. Load preprocessed train and test data
# ============================================================

train_path = "/content/final_train_processed.csv"
test_path = "/content/final_test_processed.csv"

train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

print("Training data shape:", train_df.shape)
print("Testing data shape:", test_df.shape)
```

```python
# ============================================================
# 3. Separate features and target
# ============================================================

X_train = train_df.drop("Exited", axis=1)
y_train = train_df["Exited"]

X_test = test_df.drop("Exited", axis=1)
y_test = test_df["Exited"]

print("\nX_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
```

```python
# ============================================================
# 4. Create LightGBM model
# ============================================================

lightgbm_model = lgb.LGBMClassifier(
    n_estimators=300,
    learning_rate=0.05,
    max_depth=5,
    num_leaves=31,
    subsample=0.8,
    colsample_bytree=0.8,
    class_weight="balanced",
    random_state=42,
    verbosity=-1
)
```

```python
# ============================================================
# 5. Train the model
# ============================================================

lightgbm_model.fit(
    X_train,
    y_train
)

print("\nLightGBM model trained successfully.")
```

```python
# ============================================================
# 6. Make predictions
# ============================================================

y_pred = lightgbm_model.predict(X_test)

# Probability of churn
y_prob = lightgbm_model.predict_proba(X_test)[:, 1]
```

```python

# ============================================================
# 7. Calculate evaluation metrics
# ============================================================

accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)
roc_auc = roc_auc_score(y_test, y_prob)

print("\n========== LIGHTGBM RESULTS ==========")

print(f"Accuracy  : {accuracy:.4f}")
print(f"Precision : {precision:.4f}")
print(f"Recall    : {recall:.4f}")
print(f"F1 Score  : {f1:.4f}")
print(f"ROC-AUC   : {roc_auc:.4f}")
```

```python
# ============================================================
# 8. Confusion Matrix
# ============================================================

cm = confusion_matrix(y_test, y_pred)

print("\nConfusion Matrix:")
print(cm)
```

```python
# ============================================================
# 9. Classification Report
# ============================================================

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred,
    target_names=["Retained", "Churned"]
))
```
